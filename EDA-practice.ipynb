```python
import pandas as pd
import sweetviz as sv

# 1. Load your data
df = pd.read_csv("olist_order_payments_dataset.csv")

# 2. Analyze the dataset
report = sv.analyze(df)

# 3. View the report directly inside your Jupyter Notebook cell
report.show_notebook(w="100%", h="900")
```

```python
import duckdb
import sweetviz as sv
from IPython.display import display

# 1. Run a query (for example, looking only at payments greater than 100)
query_df = duckdb.sql("""--sql
    SELECT payment_type, payment_installments, payment_value 
    FROM 'olist_order_payments_dataset.csv'
    WHERE payment_value > 100
    limit 25
""").df()  # .df() turns the query result into a Pandas DataFrame

# 2. Pass the query output straight into Sweetviz!
report = sv.analyze(query_df)

# 3. View it right inside your VS Code notebook
report.show_notebook(w="100%", h="900")

display(query_df)
```
