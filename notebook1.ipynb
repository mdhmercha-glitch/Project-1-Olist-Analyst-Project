### Write 4–6 answerable questions on sales, category mix, freight/delivery, reviews and repeat customers; name the decision-maker for each :

**Week 2 - Task 1/N [cw02-scope] — Scope and business questions**

**Skill:** Sharp, answerable questions decide which tables, grains and metrics you need. This sets up Project 1 and the G02 hiring gate, where you must explain the business questions without notes.

**Before — what to know or practice first:**
- What "answerable" means: the data can answer it with a number, a comparison or a trend
- Olist's main tables by business area: orders, order items, payments, reviews, customers, products, sellers (you have these files)
- Basic metric words: count, sum, average, median, rate, share
- A decision-maker is someone who changes something, like a seller manager or a logistics lead. A question that changes no decision is weak.

**Done well:**
- 4 to 6 questions, one per area (sales, category mix, freight/delivery, reviews, repeat customers), or fewer than 6 with clear reasons
- Each question has a number or comparison in it ("how many…", "which… by…", "what share…")
- Each question names one decision-maker and the decision it supports
- No question needs data Olist doesn't have (for example, marketing spend or profit)
- You can say which table each question likely needs

**Guide:** Don't write the questions yet. Start with one area, delivery.
